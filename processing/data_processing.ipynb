```python
import pandas as pd
import os
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
import time
```

```python
data_file = "../data/raw/art_catalog.xlsx - catalog.csv"
df = pd.read_csv(data_file)
df.head(5)
```

Selectionne uniquement les "FORM == painting""

```python
df_paint = df[df["FORM"] == "painting"]
df_paint.head(5)
```

```python
urls = list(df_paint["URL"])

urls[:5]
```

```python

output_dir = "../data/processed"
os.makedirs(output_dir, exist_ok=True) 

def download_wga_image(page_url, output_dir, sleep_time=1.5):
    img_path = page_url.replace("/html/", "/detail/")
    img_url = img_path.replace(".html", ".jpg")


    filename = os.path.join(output_dir, os.path.basename(img_url))


    if os.path.exists(filename):
        print(f"L'image existe déjà : {filename}, téléchargement ignoré.")
        return


    response = requests.get(img_url)
    if response.status_code == 200:
        with open(filename, "wb") as f:
            f.write(response.content)
        print(f"Image téléchargée : {filename}")
        time.sleep(sleep_time)  
    else:
        print("Impossible de télécharger l'image :", response.status_code, img_url)
```

```python
for url in urls:
    download_wga_image(url.strip(), output_dir)
```
